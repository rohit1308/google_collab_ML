```python
#masking with &
ages = np.array([23, 45, 17, 34, 12, 67])
incomes = np.array([30, 80, 0, 55, 0, 40])
print(ages[ages>18])
print(incomes[incomes>50])
mask=incomes>30
print(incomes[mask])
#now i want to know ages above 18 and income above 55
print(ages[(incomes>50) & (ages>=40)])
```
