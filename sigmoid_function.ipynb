```python
import numpy as np
import pandas as pd
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]

df = pd.read_csv(
    url,
    names=columns,
    skipinitialspace=True
)

df.head()
```

```python
df.shape
```

```python
df["income"].value_counts()
```

```python
df["income"] = (df["income"] == ">50K").astype(int)

df["income"].value_counts()
```

```python
from sklearn.linear_model import LinearRegression

X = df[["age"]]
y = df["income"]

model = LinearRegression()
model.fit(X, y)

pred = model.predict(X)

print(pred.min(), pred.max())
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["age"], df["income"], alpha=0.1)
plt.xlabel("Age")
plt.ylabel("Income >50K")
plt.show()
```

```python
def sigmoid(z):
  return 1 / (1 + np.exp(-z))
```

```python
print(sigmoid(100))
```
